## Ejercicio 1. Crear un árbol de categorías (20 puntos)
Construya un árbol llamado `catalogo` con raíz **Tienda**, tres categorías (**Tecnología**, **Hogar**, **Libros**) y al menos dos productos en cada categoría. Muestre el árbol completo.

**Evidencia:** estructura visual con mínimo 10 nodos (1 raíz + 3 categorías + 6 productos).

```python
# TODO: implementar el ejercicio 1
catalogo = Tree()
catalogo.create_node("Tienda", "tienda")

# Categorías
catalogo.create_node("Tecnología", "tec", parent="tienda")
catalogo.create_node("Hogar", "hogar", parent="tienda")
catalogo.create_node("Libros", "libros", parent="tienda")

# Productos
catalogo.create_node("Laptop", "laptop", parent="tec")
catalogo.create_node("Smartphone", "smartphone", parent="tec")
catalogo.create_node("Licuadora", "licuadora", parent="hogar")
catalogo.create_node("Lámpara", "lampara", parent="hogar")
catalogo.create_node("Novela", "novela", parent="libros")
catalogo.create_node("Libro de Python", "python", parent="libros")

catalogo.show()
```

# Analisis:
Modela una jerarquía de tipo “es-parte-de” (taxonomía). Matemáticamente es un grafo acíclico, conexo y dirigido, con exactamente una raíz y n−1 aristas para n nodos. Aquí n = 10 y hay 9 aristas.

Separar tag de identifier. El tag es el atributo de presentación (puede repetirse o cambiar), mientras que el identifier es la llave primaria y debe ser único. Es la misma lógica que una base de datos: si usaras la etiqueta como ID, renombrar un producto rompería las referencias. treelib lanza DuplicatedNodeIdError si se repite un ID.
Orden de creación. Un hijo solo puede crearse si su padre ya existe, lo que obliga a construir de arriba hacia abajo (top-down).

## Ejercicio 3. Modificar el árbol (20 puntos)
Sobre una **copia** del árbol anterior, cambie el nombre visible de un producto, mueva un producto a otra categoría y elimine otro producto. Muestre el árbol antes y después.

**Pista:** revise `update_node()`, `move_node()` y `remove_node()` en la documentación.

**Evidencia:** comparación de las dos estructuras.

```python
# TODO: implementar el ejercicio 3
# Copia profunda, para no alterar el árbol original
catalogo_mod = Tree(catalogo, deep=True)

print("=== ANTES ===")
catalogo_mod.show()

# 1. Cambiar el nombre visible de un producto
catalogo_mod.update_node("smartphone", tag="Smartphone 5G")

# 2. Mover un producto a otra categoría
catalogo_mod.move_node("lampara", "tec")   # Lámpara ahora en Tecnología

# 3. Eliminar un producto
catalogo_mod.remove_node("novela")

print("=== DESPUÉS ===")
catalogo_mod.show()

```

# Analisis:
Aquí se ven los tres tipos de mutación con efectos muy distintos:

update_node cambia solo un atributo (el tag). La topología no cambia, es una operación puramente semántica.
move_node cambia una arista (reasigna el padre). Si el nodo tiene descendientes, viajan con él. treelib protege la invariante de acíclico y lanza error se intenta mover un nodo debajo de su propio descendiente.
remove_node elimina el nodo y todo su subárbol. Con productos (hojas) no se nota, pero si se elimina una categoría se perderia todos sus productos.

## Ejercicio 4. Recorridos (20 puntos)
Sobre el árbol del ejercicio 1, liste los identificadores o etiquetas visitados en **profundidad (DFS)** y **anchura (BFS)**. Indique cuál recorrido visita primero todas las categorías.

**Pista:** `expand_tree(mode=Tree.DEPTH)` y `expand_tree(mode=Tree.WIDTH)`.

**Evidencia:** dos listas de recorrido y una conclusión breve (1–2 frases).

```python
# TODO: implementar el ejercicio 4
dfs = [catalogo.get_node(i).tag for i in catalogo.expand_tree(mode=Tree.DEPTH)]
bfs = [catalogo.get_node(i).tag for i in catalogo.expand_tree(mode=Tree.WIDTH)]

print("DFS (profundidad):", dfs)
print("BFS (anchura):    ", bfs)

print("\nConclusión: el recorrido en anchura (BFS) visita primero todas las "
      "categorías, porque recorre el árbol nivel por nivel; el recorrido en "
      "profundidad (DFS) baja hasta los productos de una categoría antes de "
      "pasar a la siguiente.")
```

# Analisis:
Visita la raíz, baja por la primera rama hasta el fondo y luego retrocede. Resultado típico: Tienda → Hogar → Lámpara → Licuadora → Libros → ...
BFS: visita nivel por nivel: Tienda → Hogar, Libros, Tecnología → productos.

Detalle importante que se pasa por alto: expand_tree tiene sorting=True por defecto, así que ordena los hermanos por etiqueta alfabéticamente, no por orden de inserción. Por se verá “Hogar, Libros, Tecnología” aunque se haya creado Tecnología primero. Si se necesita el orden de inserción, se usa sorting=False.

Estructura de datos subyacente: DFS usa una pila (LIFO) y BFS una cola (FIFO). Ambos son O(n) en tiempo, pero difieren en memoria: DFS ocupa O(h) (h = altura) y BFS O(w) (w = ancho máximo). En árboles anchos y poco profundos como este, DFS gasta menos memoria.
BFS visita primero todas las categorías, porque agota un nivel antes de descender.

## Ejercicio 5. Caso aplicado: servicios universitarios (20 puntos)
Diseñe un árbol de navegación para un portal universitario con raíz **Portal**, al menos cuatro servicios principales y dos trámites por servicio.

Implemente una función `ruta_hasta_raiz(arbol, identificador)` que devuelva la ruta desde un trámite hasta la raíz (en orden raíz → trámite). Pruebe la función con dos trámites diferentes.

**Evidencia:** árbol visual y dos rutas completas.

```python
# TODO: implementar el ejercicio 5
portal = Tree()
portal.create_node("Portal", "portal")

# 4 servicios principales
portal.create_node("Admisiones", "adm", parent="portal")
portal.create_node("Biblioteca", "bib", parent="portal")
portal.create_node("Bienestar", "bien", parent="portal")
portal.create_node("Registro Académico", "reg", parent="portal")

# 2 trámites por servicio
portal.create_node("Inscripción a programas", "adm1", parent="adm")
portal.create_node("Consulta de requisitos", "adm2", parent="adm")
portal.create_node("Préstamo de libros", "bib1", parent="bib")
portal.create_node("Reserva de salas", "bib2", parent="bib")
portal.create_node("Solicitud de becas", "bien1", parent="bien")
portal.create_node("Cita con psicología", "bien2", parent="bien")
portal.create_node("Certificado de notas", "reg1", parent="reg")
portal.create_node("Matrícula de materias", "reg2", parent="reg")

portal.show()

def ruta_hasta_raiz(arbol, identificador):
    """Devuelve la ruta raíz → trámite como lista de etiquetas."""
    # rsearch recorre desde el nodo hacia la raíz; se invierte el orden
    ruta = [arbol.get_node(i).tag for i in arbol.rsearch(identificador)]
    return ruta[::-1]

# Pruebas con dos trámites distintos
print(" → ".join(ruta_hasta_raiz(portal, "bib2")))
print(" → ".join(ruta_hasta_raiz(portal, "reg1")))
```

# Analisis:
Un árbol de navegación equivale a una ruta de migas de pan (breadcrumb), un patrón de UX muy común. Con 1 raíz + 4 servicios + 8 trámites son 13 nodos.

La función: como cada nodo tiene un único padre, existe un solo camino entre cualquier nodo y la raíz, por lo que la solución es determinista. rsearch sube por la cadena de padres, así que cuesta O(h) y no O(n). La inversión [::-1] cuesta O(h) adicional. En total es O(h), muy eficiente para jerarquías profundas.
